```python

import getpass
from google import genai
from concurrent.futures import ThreadPoolExecutor

# Securely prompt for your Google AI Studio API key if not already set
api_key = getpass.getpass("Enter your Google AI Studio API Key: ")

# Initialize the official Gemini API client
client = genai.Client(api_key=api_key)

# Hardcoded registry of expert personas that Professor Synapse can morph into
PERSONAS = {
    "synapse_conductor": (
        "You are Professor Synapse, a strategic enterprise architect and conductor. "
        "Your role is to interview the user, analyze intent, and map out the execution plan."
    ),
    "python_architect": (
        "You are an elite Python Systems Architect. Focus strictly on clean code, "
        "type hinting, error handling, and runtime safety."
    ),
    "skeptical_critic": (
        "You are a reality-oriented critic. Find logical flaws, missing base rates, "
        "and hidden failure points in the proposed design."
    )
}

def invoke_expert(persona_key: str, task_prompt: str) -> str:
    """Professor Synapse temporarily morphs its identity into the specified expert profile."""
    instruction = PERSONAS.get(persona_key, PERSONAS["synapse_conductor"])
    print(f"\n[Swarm Execution] Professor Synapse morphing into: [{persona_key.upper()}]...")

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=task_prompt,
        config={
            "system_instruction": instruction,
            "temperature": 0.2,
        },
    )
    return response.text

def run_parallel_swarm(task: str):
    """Spawns parallel threads to let multiple expert personas review the task concurrently."""
    experts = ["python_architect", "skeptical_critic"]
    results = {}

    print(f"Analyzing Task: '{task}'\n" + "-"*40)
    with ThreadPoolExecutor(max_workers=2) as executor:
        futures = {exp: executor.submit(invoke_expert, exp, task) for exp in experts}
        for exp, future in futures.items():
            results[exp] = future.result()

    return results

# Test execution with a sample architecture problem
project_test = "We are designing an automated data-scraping script. What are the key architectural risks?"
swarm_results = run_parallel_swarm(project_test)

for expert, output in swarm_results.items():
    print(f"\n=== Output from [{expert.upper()}] ===")
    print(output)
```
